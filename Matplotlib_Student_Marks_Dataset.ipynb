# Matplotlib Student Marks Dataset
This notebook contains the sample student marks dataset and practice graphs.

**Dataset:** Maths, Physics, Chemistry and Biology marks, with an error value for each subject.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Subject': ['Maths', 'Physics', 'Chemistry', 'Biology'],
    'Marks': [85, 75, 90, 80],
    'Error': [3, 4, 2, 5]
}

df = pd.DataFrame(data)
df
```

## 1. Title and Axis Formatting

```python
plt.figure(figsize=(7, 4), dpi=100)
plt.plot(df['Subject'], df['Marks'], color='blue', linewidth=2, marker='o')
plt.title('Student Marks', fontsize=14, fontweight='bold')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.xticks(rotation=0)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Line Graph with Formatting Properties

```python
plt.figure(figsize=(7, 4))
plt.plot(df['Subject'], df['Marks'], color='green', linewidth=2,
         linestyle='--', marker='o', markersize=8,
         markerfacecolor='yellow', markeredgecolor='black',
         alpha=0.8, label='Marks')
plt.title('Student Marks - Line Graph')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Vertical Bar Graph

```python
plt.figure(figsize=(7, 4))
bars = plt.bar(df['Subject'], df['Marks'], color='skyblue', width=0.6,
               edgecolor='black', linewidth=1, alpha=0.9,
               label='Marks', hatch='//',
               yerr=df['Error'], capsize=5)
plt.bar_label(bars, padding=3)
plt.title('Student Marks - Bar Graph')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.grid(axis='y', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Horizontal Bar Graph

```python
plt.figure(figsize=(7, 4))
plt.barh(df['Subject'], df['Marks'], color='orange', edgecolor='black', label='Marks')
plt.title('Student Marks - Horizontal Bar Graph')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Bar Graph Using the Bottom Property

```python
plt.figure(figsize=(7, 4))
plt.bar(df['Subject'], df['Marks'], bottom=5, color='mediumpurple', edgecolor='black')
plt.title('Student Marks with Bottom Property')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.tight_layout()
plt.show()
```

## 6. Bar Graph with Custom Tick Labels

```python
custom_labels = ['M', 'P', 'C', 'B']
plt.figure(figsize=(7, 4))
plt.bar(range(len(df)), df['Marks'], tick_label=custom_labels, color='teal')
plt.title('Student Marks with Custom Tick Labels')
plt.xlabel('Subject Initials')
plt.ylabel('Marks')
plt.tight_layout()
plt.show()
```

## 7. Pie Chart (Reduced Size)

```python
plt.figure(figsize=(4, 4))
plt.pie(df['Marks'], labels=df['Subject'], autopct='%1.1f%%',
        startangle=90, explode=(0.05, 0, 0, 0), shadow=True,
        radius=0.6, textprops={'fontsize': 9},
        wedgeprops={'edgecolor': 'white', 'linewidth': 1},
        pctdistance=0.7, labeldistance=1.1)
plt.title('Marks Distribution', fontsize=12)
plt.legend(df['Subject'], loc='best', fontsize=7)
plt.tight_layout()
plt.show()
```

## 8. Save a Graph

```python
plt.figure(figsize=(7, 4))
plt.bar(df['Subject'], df['Marks'], color='cornflowerblue')
plt.title('Student Marks')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.tight_layout()
plt.savefig('student_marks_graph.png', dpi=300, bbox_inches='tight')
plt.show()
```
